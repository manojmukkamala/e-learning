```python
def twoSum(nums, target):
    """
    :type nums: List[int]
    :type target: int
    :rtype: List[int]
    """
    # print(nums)
    # print(target)
    l = len(nums)
    # print(l)
    if l <= 1: 
        print("Length of array should be more than 1")
    else:
        
        # if two elements are next to each other, the following logic works, otherwise not
        # for i in range(l-1):
        #     print(i)
        #     t = nums[i] + nums[i+1]
        #     if t == target:
        #         return [i, i+1]

        # brute force
        # for i in range(l):
        #     # print(i)
        #     for j in range(i+1, l):
        #         # print(i, j)
        #         # print(nums[i], nums[j])
        #         t = nums[i] + nums[j]
        #         if t == target:
        #             return [i, j]

        d = {}

        for i in range(l):
            d[nums[i]] = i

        for i in range(l):
            c = target - nums[i]
            if (c in d) and (d[c] != i):
                return [i, d[c]]
                

nums = [2,11,15,0,1,3,7,12]
target = 9

# nums = [3,2,4]
# target = 6

# nums = [3,3]
# target = 6

twoSum(nums, target)
```
