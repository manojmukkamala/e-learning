### Bernouli Variable

Lets toss a __fair__ coin 100000 times and note down the outcomes.

A coin toss results in 2 possible outcomes: 
- Heads with P(H) = 0.5 
- Tails with P(T) = 0.5

Lets define a bernouli variable ```bernouli``` to capture all the 100000 outcomes. A bernouli can take two possible values: 
- Success (1)
- Failure (0)

---

````python
bernouli = np.random.randint(0, 2, 100000)
````

---

Let ```N``` be the total number of flips (100000).

Let ```p``` be the proportion of heads (successes) in the population.

---

````python
N = len(bernouli)
p = sum(bernouli == 1)/N
````

````python
plt.hist(bernouli, ec = 'k')
plt.show()
````

---

#### Population Parameters

__Mean__ - The mean of bernouli variable is ```p```, the proportion of heads.

__Variance__ - The variance of bernouli variable is ```p*(1-p)```.

__Standard Deviation__ - The standard deviation of bernouli variable is ```sqrt(p*(1-p))```.

---

````python
print("Mean of Bernouli (Statistics):", p)
print("Mean of Bernouli (Numpy):", np.mean(bernouli))
````

````python
print("Variance of Bernouli (Statistics):", p*(1-p))
print("Variance of Bernouli (Numpy):", np.var(bernouli))
````

````python
print("Std Dev of Bernouli (Statistics):", np.sqrt(p*(1-p)))
print("Std Dev of Bernouli (Numpy):", np.std(bernouli))
````

### Binomial Variable

From the above population, lets randomly pick a 50 samples and count the number of heads (successes) in those 50 samples.

Lets define a binomial variable ```X``` to capture the __SUM__ of number of heads (successes) in those 50 flips.

Lets repeat the same experiment (picking 50 random samples, counting the number of heads) for 1000 times and capture the results in variable X.

---

```
EXPERIMENT_1 = SAMPLE_11 + SAMPLE_12 + SAMPLE_13 .... + SAMPLE_150
EXPERIMENT_2 = SAMPLE_21 + SAMPLE_22 + SAMPLE_23 .... + SAMPLE_250
.
.
.
.
EXPERIMENT_1000 = SAMPLE_10001 + SAMPLE_10002 + SAMPLE_10003 ... + SAMPLE_100050
```

---

Let ```n``` be the sample size (50).

Let ```exps``` be the number of experiments (1000).

````python
n = 50
exps = 1000
sample_index = np.random.randint(0, N, (n*exps))
````

````python
X = np.sum(np.reshape(bernouli[sample_index], (1000, 50)), axis = 1)
````

````python
plt.hist(X, ec = 'k')
plt.show()
````

---

#### Sample Statistics

__Mean__ - The mean of binomial variable is ```n * p```.

__Variance__ - The variance of binomial variable is ```n * p*(1-p)```.

__Standard Deviation__ - The standard deviation of binomial variable is ```sqrt(n * p*(1-p))```.

---

````python
print("Mean of X (Statistics):", n * p)
print("Mean of X (Numpy):", np.mean(X))
````

````python
print("Variance of X (Statistics):", n * p * (1-p))
print("Variance of X (Numpy):", np.var(X))
````

````python
print("Std Dev of X (Statistics):", np.sqrt(n * p * (1-p)))
print("Std Dev of X (Numpy):", np.std(X))
````

### Central Limit Theorem

Central Limit Theorem states that the sampling distribution of the sample means approximates __Normal distribution.__

In the above example, we computed the sum of successes in each experiment and defined a binomial variable X to capture these coefficients. 

Lets define a variable ```p_hat``` to capture the __MEAN__ number of heads (successes) in those 50 flips.

Lets repeat the same experiment (picking 50 random samples, count the average number of heads) for 1000 times and capture the results in variable p_hat.

---

```
EXPERIMENT_1 = (SAMPLE_11 + SAMPLE_12 + SAMPLE_13 .... + SAMPLE_150)/50
EXPERIMENT_2 = (SAMPLE_21 + SAMPLE_22 + SAMPLE_23 .... + SAMPLE_250)/50
.
.
.
.
EXPERIMENT_1000 = (SAMPLE_10001 + SAMPLE_10002 + SAMPLE_10003 ... + SAMPLE_100050)/50
```

---

````python
p_hat = X/n
````

````python
plt.hist(p_hat, ec = 'k')
plt.show()
````

---

According to Central Limit Theorem, as the sample size n increases, the sampling distribution of sample means will have a smaller standard deviation.

Let's define two key definitions:

__Skew__: 

- Perfect Normal Distribution: skew = 0
- Large Right Tail (Right Skewed): skew is +ve.
- Large Left Tail (Left Skewed): skew is -ve.

__Kurtosis__:

- Fatter tails & Pointy Peak: Positive Kurtosis
- Smaller tails & smoother Peak: Negative Kurtosis

---

---

#### Sample Statistics

According to Central Limit Theorem:

__Mean__ - The mean of sampling distribution of sample means is ```p```. Note that the mean of sampling distribution of sample mean is equal to population mean.

__Variance__ - The variance of sampling distribution of sample means is equal to population variance divided by sample size ```(1/n) * p*(1-p)```.

__Standard Deviation__ - The standard deviation of sampling distribution of sample means is equal to population standard deviation divided by square root of sample size ```sqrt((1/n) * p*(1-p))```.

---

````python
print("Mean of p_hat (Statistics):", p)
print("Mean of p_hat (Numpy):", np.mean(p_hat))
````

````python
print("Variance of p_hat (Statistics):", (1/n) * p * (1-p))
print("Variance of p_hat (Numpy):", np.var(p_hat))
````

````python
print("Std Dev of p_hat (Statistics):", np.sqrt((1/n) * p * (1-p)))
print("Std Dev of p_hat (Numpy):", np.std(p_hat))
````

### Confidence Intervals

Lets pick one sample from the sample proportions we computed earlier.

Assume that we do not have any other details about the population and lets construct the confidence intervals.

````python
p_hat_1 = p_hat[0]
print("Sample Proportion:", p_hat_1)
````

---

The above sample represents just one sample from the sampling distribution of sample means. 

According to central limit theorem, the sampling distribution of sample means follws a normal distribution. So, this sample is part of a distribution that approximates normal distribution.

We would like to find out how close the sample proportion is to the population proportion. In other words, what is the probability that p_hat_1 is within 2 standard deviations of p?

We know that, for a normal dist, approx 95% of area is within 2 standard deviations.

Let's put it the other way: There is 95% probability that p_hat_1 is within 2 standard deviations of p. This is equivalent to say that there is 95% chance that p is within 2 standard deviations of p_hat_1.

---

The mean of the sampling distribution of sample means is equal to population mean ```p```. 

The standard deviation of sampling distribution of sample means is equal to population standard deviation divided by square root of sample size ```sqrt((1/n) * p*(1-p))```.

---

Now, we should compute the standard deviation of sampling distribution of sample means to build the confidence intervals.

However, as the standard deviation of sampling distribution of sample means requires population proportion p, and in this scenario we don't know the population proportion p beforehand, we cannot compute the standard deviation of sampling distribution of sample means.

---

Let's define a new statistic, Standard Error, to compute the standard deviation of sampling distribution of sample proportions.

```Standard Error = sqrt((1/n) * p_hat_1 * (1 - p_hat))```

````python
std_error = np.sqrt((1/n) * p_hat_1 * (1 - p_hat_1))
print("Standard Error:", std_error)
````

````python
margin_of_error = 2 * std_error
print("Margin of Error:", margin_of_error)

(ll, ul) = p_hat_1 - margin_of_error, p_hat_1 + margin_of_error
print("Confidence Interval:", (ll, ul))
````

````python
lo, up = p_hat - (2 * np.sqrt(p_hat * (1 - p_hat) * (1/n))), p_hat + (2 * np.sqrt(p_hat * (1 - p_hat) * (1/n)))
````

````python
sum((lo < p) == (p < up))/len(p_hat)
````
