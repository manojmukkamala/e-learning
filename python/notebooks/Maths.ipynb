```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
%matplotlib inline
```

```
X = np.random.randint(0, 2, 1000000)
X.shape
```

```
p = sum(X)/len(X)
print("Population Proportion p:", p)
```

```
sample_size = 100
num_of_samples = 1000
```

```
Y = np.array([])
for i in range(num_of_samples):
    Y = np.append(Y, np.sum(np.random.choice(X, size = sample_size)))
```

```
p_hat = Y/sample_size
```

```
plt.hist(p_hat)
plt.title('Sampling Distribution of Sample Proportions for Sample Size = 100');
```

```
np.mean(p_hat)
```

```
np.std(p_hat)
```

```
np.sqrt((p*(1-p))/sample_size)
```

Confidence Intervals

```
phat = p_hat[2]
phat
```

```
std_error = np.sqrt((p_hat[2]*(1-p_hat[2]))/sample_size)
```

```
p_hat[2]-2*std_error, p_hat[2]+2*std_error
```

```
std_error = np.sqrt((p_hat * (1 - p_hat))/sample_size)
```

```
np.mean(((p_hat - 2*std_error) <= p) & (p <= (p_hat + 2*std_error)))
```

```
plt.figure(figsize=(16, 20))
plt.scatter((p_hat - 2*std_error), *[range(0, num_of_samples)], marker = "|")
plt.scatter((p_hat + 2*std_error), *[range(0, num_of_samples)], marker = "|")
plt.plot([p for i in range(num_of_samples)], *[range(0, num_of_samples)])
```
